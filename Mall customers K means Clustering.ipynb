```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
d=pd.read_csv("/content/drive/MyDrive/MLT /Mall_Customers.csv")
d.head(5)
```

```python
d.isnull().sum()
```

```python
plt.scatter(d['Annual_Income_(k$)'],d['Spending_Score'])
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score')
```

```python
sns.pairplot(d)
```

```python
q=d.iloc[:,3:].values
q
```

```python
cs=[]
for i in range(1,11):
  kmeans=KMeans(n_clusters=i,init='k-means++',random_state=42)
  kmeans.fit(q)
  cs.append(kmeans.inertia_)
plt.plot(cs)
```

```python
kmeans=KMeans(n_clusters=6,init='k-means++',random_state=42)
y_kmeans=kmeans.fit_predict(q)
y_kmeans
```

```python
kmeans.inertia_
```

```python
kmeans.labels_
```

```python
plt.scatter(q[y_kmeans==0,0],q[y_kmeans==0,1],color='yellow',label='cluster1')
plt.scatter(q[y_kmeans==1,0],q[y_kmeans==1,1],color='blue',label='cluster2')
plt.scatter(q[y_kmeans==2,0],q[y_kmeans==2,1],color='green',label='cluster3')
plt.scatter(q[y_kmeans==3,0],q[y_kmeans==3,1],color='cyan',label='cluster4')
plt.scatter(q[y_kmeans==4,0],q[y_kmeans==4,1],color='magenta',label='cluster5')
plt.scatter(q[y_kmeans==5,0],q[y_kmeans==5,1],color='red',label='cluster6')
plt.scatter(kmeans.cluster_centers_[:,0],kmeans.cluster_centers_[:,1],color='black',label='centroid',s=300,marker='+')
plt.legend()
plt.show()
```
